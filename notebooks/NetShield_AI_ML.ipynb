```python
#combining all 8 csv files
import pandas as pd
import glob
import os

dataset_path = "/content/drive/MyDrive/Netshield-Ai/data/raw/extracted/MachineLearningCVE"

csv_files = glob.glob(os.path.join(dataset_path, "*.csv"))

dataframes = []

for file in csv_files:
    print("Loading:", os.path.basename(file))

    df = pd.read_csv(file, low_memory=False)
    dataframes.append(df)

    print("Rows:", len(df))

dataset = pd.concat(dataframes, ignore_index=True)

print("\nCombined Dataset Shape:", dataset.shape)

output_path = "/content/drive/MyDrive/Netshield-Ai/data/combined_dataset.csv"

dataset.to_csv(output_path, index=False)

print("\nCombined dataset saved successfully!")
```

```python
#data quality inspection(checking missing and infinite values in training data)
import pandas as pd
import numpy as np

dataset_path = "/content/drive/MyDrive/Netshield-Ai/data/combined_dataset.csv"

df = pd.read_csv(dataset_path, low_memory=False)

print("Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nInfinite Values:")
print(np.isinf(df.select_dtypes(include=np.number)).sum().sum())

print("\nData Types:")
print(df.dtypes.value_counts())
```

```python
#cleaning dataset
import pandas as pd
import numpy as np

# Load dataset
path = "/content/drive/MyDrive/Netshield-Ai/data/combined_dataset.csv"
df = pd.read_csv(path, low_memory=False)

print("Original Shape:", df.shape)

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Replace infinity with NaN
df.replace([np.inf, -np.inf], np.nan, inplace=True)

# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Remove rows with missing target labels
df.dropna(subset=["Label"], inplace=True)

print("\nAfter Cleaning:")
print("Dataset Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())

# Save cleaned dataset
output_path = "/content/drive/MyDrive/Netshield-Ai/data/cleaned_dataset.csv"

df.to_csv(output_path, index=False)

print("\nCleaned dataset saved successfully!")
```

```python
#visualise attack chart
import pandas as pd
import matplotlib.pyplot as plt

path = "/content/drive/MyDrive/Netshield-Ai/data/cleaned_dataset.csv"

df = pd.read_csv(path, usecols=["Label"])

counts = df["Label"].value_counts()

plt.figure(figsize=(12, 6))
counts.plot(kind="bar")

plt.title("Network Traffic Distribution - NetShield AI")
plt.xlabel("Traffic Category")
plt.ylabel("Number of Records")
plt.xticks(rotation=70)
plt.tight_layout()
plt.show()
```

```python
#logarithmic scale (bcz this dataset is imbalance and chart is difficult to understand)
import pandas as pd
import matplotlib.pyplot as plt

path = "/content/drive/MyDrive/Netshield-Ai/data/cleaned_dataset.csv"

df = pd.read_csv(path, usecols=["Label"])

counts = df["Label"].value_counts().sort_values()

plt.figure(figsize=(12, 8))

counts.plot(kind="barh")

plt.xscale("log")
plt.title("NetShield AI - Network Traffic Distribution")
plt.xlabel("Number of Records (Log Scale)")
plt.ylabel("Traffic Category")

plt.tight_layout()
plt.show()
```

```python
#identifying missing values
import pandas as pd

path = "/content/drive/MyDrive/Netshield-Ai/data/cleaned_dataset.csv"

missing_counts = {}

for chunk in pd.read_csv(path, chunksize=100000, low_memory=False):
    counts = chunk.isnull().sum()

    for column, count in counts.items():
        missing_counts[column] = missing_counts.get(column, 0) + count

missing = pd.Series(missing_counts)
missing = missing[missing > 0].sort_values(ascending=False)

print("Features containing missing values:")
print(missing)

print("\nTotal missing values:", missing.sum())
```
